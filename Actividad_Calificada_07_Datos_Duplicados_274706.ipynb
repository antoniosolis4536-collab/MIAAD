```python
import pandas as pd

# Carga del dataset Hotel Booking Demand desde GitHub (TidyTuesday, 2020-02-11)
url = 'https://raw.githubusercontent.com/rfordatascience/tidytuesday/master/data/2020/2020-02-11/hotels.csv'
hoteles = pd.read_csv(url)

# Dimensiones y primeras filas
print('Dimensiones:', hoteles.shape)
hoteles.head()
```

## Actividad 1. Diagnóstico inicial (20 puntos)

### 1.1 Conteo de duplicados exactos (5 pts)

Calcula el número de filas marcadas como duplicadas en `hoteles` considerando **todas** las columnas, con `keep='first'`, `keep='last'` y `keep=False`. Imprime los tres resultados con un texto que explique qué significa cada uno.

```python
# Escribe tu código aquí
# Conteo de duplicados exactos usando keep
dup_first = hoteles.duplicated(keep='first').sum()
dup_last = hoteles.duplicated(keep='last').sum()
dup_false = hoteles.duplicated(keep=False).sum()

# Imprimir los resultados
print("Número de filas duplicadas (keep='first') no contempla el primero:\n", dup_first)
print("Número de filas duplicadas (keep='last') no contempla el ultimo:\n", dup_last)
print("Número de filas duplicadas (keep=False) contempla todas:\n", dup_false)
```

### 1.2 Magnitud del problema (5 pts)

Con los resultados de 1.1, calcula e imprime:

- El **porcentaje** de filas sobrantes respecto al total.
- El número de **grupos repetidos** (combinaciones de valores que aparecen más de una vez).

```python
# Escribe tu código aquí
porcentaje_sobrante = dup_first / len(hoteles) * 100
print(f"Porcentaje de filas sobrantes: {porcentaje_sobrante:.2f}%")

# numero de grupos repetidos
conteo_combinaciones = hoteles.value_counts(dropna=False)
grupos_repetidos = conteo_combinaciones[conteo_combinaciones > 1].sum()
print(f"Número de grupos repetidos: {grupos_repetidos}")
```

### 1.3 Duplicados por hotel (5 pts)

Determina **cuántas** filas sobrantes (`keep='first'`) hay en cada hotel y qué **porcentaje** representan respecto al total de reservas de ese hotel.

Sugerencia: crea una copia de `hoteles`, agrega una columna booleana `es_duplicado` con el resultado de `duplicated()` y agrupa por `hotel`.

```python
# Escribe tu código aquí
hoteles_copia = hoteles.copy()
hoteles_copia['es_duplicado'] = hoteles_copia.duplicated(keep='first')

resumen_hoteles = hoteles_copia.groupby('hotel')['es_duplicado'].agg(['sum', 'count'])
resumen_hoteles['porcentaje'] = resumen_hoteles['sum'] / resumen_hoteles['count'] * 100
print(resumen_hoteles)
```

## Actividad 3. Impacto en el análisis (25 puntos)

### 3.1 Versión sin duplicados (5 pts)

Crea `hoteles_limpio` eliminando los duplicados exactos, conservando la primera aparición y **renumerando el índice**. Imprime el número de filas antes y después y cuántas se eliminaron.

```python
# Escribe tu código aquí
filas_antes = len(hoteles)
hoteles_limpio = hoteles.drop_duplicates(keep='first', ignore_index=True)
filas_despues = len(hoteles_limpio)

print(f"Filas antes: {filas_antes}")
print(f"Filas después: {filas_despues}")
print(f"Filas eliminadas: {filas_antes - filas_despues}")
```

### 3.2 Métricas por hotel (8 pts)

Para `hoteles` y para `hoteles_limpio`, calcula por hotel con `groupby()`:

- Número de reservas (`.size()`).
- Tasa de cancelación (media de `is_canceled`).
- Tarifa diaria promedio (media de `adr`).

Nota: para calcular la media de varias columnas a la vez, selecciona una lista de columnas: `df.groupby('hotel')[['col1', 'col2']].mean()`.

Guarda las medias en dos `DataFrame`, `resumen_original` y `resumen_limpio`, con `hotel` como columna (usa `reset_index()`).

```python
# Escribe tu código aquí
resumen_original = hoteles.groupby('hotel').agg(
    reservas=('hotel', 'size'),
    tasa_cancelacion=('is_canceled', 'mean'),
    tarifa_promedio=('adr', 'mean')
).reset_index()

resumen_limpio = hoteles_limpio.groupby('hotel').agg(
    reservas=('hotel', 'size'),
    tasa_cancelacion=('is_canceled', 'mean'),
    tarifa_promedio=('adr', 'mean')
).reset_index()

print("Resumen original:")
print(resumen_original)

print("\nResumen limpio:")
print(resumen_limpio)
```

### 3.3 Tabla comparativa (6 pts)

Agrega a cada resumen una columna `version` (`'original'` y `'sin duplicados'`) y únelos con `pd.concat()` en un `DataFrame` llamado `comparacion`, con índice renumerado y ordenado por `hotel`, de modo que las dos versiones de cada hotel queden juntas.

```python
# Escribe tu código aquí
resumen_original['version'] = 'original'
resumen_limpio['version'] = 'sin duplicados'

comparacion = pd.concat([resumen_original, resumen_limpio], ignore_index=True)
comparacion = comparacion.sort_values(by='hotel').reset_index(drop=True)
comparacion
```
