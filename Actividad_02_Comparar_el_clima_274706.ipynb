**Define una lista ciudades con al menos tres ciudades (nombre, latitud, longitud), distintas a Ciudad Juárez (la usada en el Bloque 2).**

```python
# Ciudades visitadas por Miguelito
ciudades = [
    ("Ciudad de Mexico", 19.43, -99.13),
    ("Guadalajara MX", 20.67, -103.34),
    ("Monterrey MX" , 25.67, -103.35),
    ("Tijuana MX", 32.51, -117.02),
    ("Queretaro MX", 20.58, -100.39),
    ("Atlanta GA", 33.74, -84.38),
    ("Syracuse NY", 43.04, -76.14),
    ("Greene NY", 42.33, -75.77),
    ("Hornell NY", 42.32, -77.66),
    ("Toronto CAN", 43.65, -79.38),
]
```

**Para cada ciudad, consulta el endpoint https://api.open-meteo.com/v1/forecast con el parámetro current_weather=true, usando timeout y manejo de errores (try/except con raise_for_status()), igual que en el Bloque 2.**

**De cada respuesta, extrae la temperatura y la velocidad del viento actuales (dentro de datos["current_weather"]).**

```python
import requests
import pandas as pd

endpoint = "https://api.open-meteo.com/v1/forecast"

info_clima = []

for nombre, latitud, longitud in ciudades:
    params = {
        "latitude": latitud,
        "longitude": longitud,
        "current_weather": True
    }
    try:
        response = requests.get(endpoint, params=params, timeout=30)
        response.raise_for_status()
        data = response.json()
        origen_respuesta = "Open-Meteo (consulta en linea)"
        print("status_code:", response.status_code)

        clima_actual = data["current_weather"]

        info_clima.append({
            "Ciudad": nombre,
            "Temperatura": clima_actual["temperature"],
            "Velocidad del Viento": clima_actual["windspeed"]
        })
    except requests.RequestException as error:
      print(f"no fue posible consultar el clima de: {nombre}:", error)
      info_clima.append({
            "Ciudad": nombre,
            "Temperatura": None,
            "Velocidad del Viento": None
        })
```

**Estructura la información como una lista de diccionarios llamada info_clima.**

```python
info_clima
```

**Convierte info_clima en un DataFrame llamado df_clima_ciudades.**

```python

df_clima_ciudades = pd.DataFrame(info_clima)
df_clima_ciudades
```

**Verifica el resultado: dimensiones, columnas, y valores faltantes.**

```python
print("Dimensiones del DataFrame:")
print(df_clima_ciudades.shape)

print("\nColumnas del DataFrame:")
print(df_clima_ciudades.columns)

print("\nValores faltantes en cada columna:")
print(df_clima_ciudades.isnull().sum())
```
