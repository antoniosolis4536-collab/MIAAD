**Descarga el dataset prathamtripathi/drug-classification en la carpeta datos_drug/.**

```python
dataset_id = "prathamtripathi/drug-classification"

!kaggle datasets download -d {dataset_id} -p datos_drug/ --unzip
```

**Carga el archivo .csv descargado en un DataFrame llamado df_drug.**

```python
import glob
import pandas as pd

archivos_csv = glob.glob("datos_drug/*.csv")
print(archivos_csv)

if archivos_csv:
    df_drug = pd.read_csv(archivos_csv[0])
    print(df_drug.shape)
    df_drug.head()
```

**Verifica el resultado: dimensiones, columnas, tipos de datos, y valores faltantes por columna.**

```python
print("Dimensiones:", df_drug.shape)
print()
df_drug.info()
print()
print("Valores faltantes por columna:", df_drug.isna().sum())
```

**Completa el diccionario ficha_drug con la información de procedencia del dataset (fuente, fecha de descarga, licencia — revisa la página del dataset en kaggle.com para este último dato).**

```python
from datetime import datetime, timezone

ficha_drug = {
    "fuente": f"Kaggle - {dataset_id}",
    "url": f"https://www.kaggle.com/datasets/{dataset_id}",
    "fecha_descarga_utc": datetime.now(timezone.utc).isoformat(),
    "licencia": "CC0: Public Domain",
    "archivo_leido": archivos_csv[0],
    "Dimensiones": df_drug.shape
}

ficha_drug
```
